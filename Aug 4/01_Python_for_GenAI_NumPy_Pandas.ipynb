## 1. Python Refresher

A quick recap of the Python building blocks we will rely on for the rest of the course: variables and
data types, control flow, functions, comprehensions, and a minimal class example.


```python
# Variables and core data types
name = "GenAI Learner"
age = 25
height_m = 1.75
is_enrolled = True
skills = ["python", "numpy", "pandas"]
profile = {"name": name, "age": age, "skills": skills}

print(type(name), type(age), type(height_m), type(is_enrolled), type(skills), type(profile))
print(profile)
```

```python
# Control flow: if / for / while
for skill in skills:
    if skill.startswith("p"):
        print(skill, "-> starts with p")
    else:
        print(skill, "-> does not start with p")

count = 0
while count < 3:
    print("while loop iteration:", count)
    count += 1
```

```python
# Functions: default args, *args, **kwargs, lambda
def describe_person(name, age=18, **extra_info):
    base = f"{name} is {age} years old."
    if extra_info:
        details = ", ".join(f"{k}={v}" for k, v in extra_info.items())
        base += f" Extra: {details}"
    return base

print(describe_person("Ahmed", age=30, role="Instructor", city="Bengaluru"))

def total(*numbers):
    return sum(numbers)

print("total:", total(1, 2, 3, 4, 5))

square = lambda x: x ** 2
print("square(7) =", square(7))
```

```python
# Comprehensions: list, dict, set
numbers = list(range(1, 11))

squares = [n ** 2 for n in numbers]
even_squares = [n ** 2 for n in numbers if n % 2 == 0]
square_lookup = {n: n ** 2 for n in numbers}
unique_remainders = {n % 3 for n in numbers}

print("squares:", squares)
print("even_squares:", even_squares)
print("square_lookup:", square_lookup)
print("unique_remainders:", unique_remainders)
```

```python
# A minimal class example
class Student:
    """Represents a student on the course."""

    def __init__(self, name, scores):
        self.name = name
        self.scores = scores

    def average_score(self):
        return sum(self.scores) / len(self.scores)

    def __repr__(self):
        return f"Student(name={self.name!r}, average={self.average_score():.2f})"


students = [Student("Asha", [88, 92, 79]), Student("Ravi", [65, 70, 72])]
for s in students:
    print(s)
```

## 2. NumPy

NumPy is the core library for fast numerical computation in Python. Pandas, scikit-learn and most
machine learning libraries are built on top of it.


```python
import numpy as np

# Creating arrays
a = np.array([1, 2, 3, 4, 5])
b = np.arange(0, 10, 2)          # start, stop, step
c = np.linspace(0, 1, 5)         # 5 evenly spaced points between 0 and 1
zeros = np.zeros((2, 3))
ones = np.ones((3, 2))
identity = np.eye(3)

print("a:", a)
print("b:", b)
print("c:", c)
print("zeros:\n", zeros)
print("identity:\n", identity)
print("a dtype/shape:", a.dtype, a.shape)
```

```python
# Indexing, slicing and boolean masking
matrix = np.arange(1, 13).reshape(3, 4)
print("matrix:\n", matrix)

print("first row:", matrix[0])
print("second column:", matrix[:, 1])
print("sub-matrix:\n", matrix[0:2, 1:3])

mask = matrix % 2 == 0
print("even mask:\n", mask)
print("even values:", matrix[mask])
```

```python
# Vectorised operations and broadcasting
x = np.array([1, 2, 3])
y = np.array([10, 20, 30])

print("x + y:", x + y)
print("x * 2:", x * 2)
print("dot product:", np.dot(x, y))

# Broadcasting: adding a (3,) vector to every row of a (3,3) matrix
grid = np.ones((3, 3))
print("grid + x:\n", grid + x)
```

```python
# Aggregations and random numbers
rng = np.random.default_rng(seed=42)
sample = rng.normal(loc=50, scale=10, size=1000)

print("mean:", sample.mean())
print("std:", sample.std())
print("min/max:", sample.min(), sample.max())
print("argmax index:", sample.argmax())
```

## 3. Pandas

Pandas gives us `Series` (1-D labelled arrays) and `DataFrame` (2-D labelled tables), which are the
standard way to work with structured/tabular data in Python.


```python
import pandas as pd

# Series and DataFrame creation
s = pd.Series([10, 20, 30, 40], name="scores")
print(s)

df = pd.DataFrame({
    "name": ["Asha", "Ravi", "Meera", "Kiran"],
    "age": [23, 25, 22, 28],
    "department": ["Data", "AI", "Data", "AI"],
    "score": [88, 74, 91, 67],
})
df
```

```python
# Writing to and reading from CSV
df.to_csv("students.csv", index=False)
loaded_df = pd.read_csv("students.csv")
loaded_df
```

```python
# Indexing and filtering
print(df.loc[0, "name"])          # label based
print(df.iloc[0, 0])              # position based

high_scorers = df[df["score"] >= 80]
high_scorers
```

```python
# Groupby and aggregation
dept_summary = df.groupby("department")["score"].agg(["mean", "min", "max", "count"])
dept_summary
```

```python
# Merge / join and sorting
extra_info = pd.DataFrame({
    "name": ["Asha", "Ravi", "Meera", "Kiran"],
    "city": ["Bengaluru", "Hyderabad", "Chennai", "Pune"],
})

merged = pd.merge(df, extra_info, on="name")
sorted_df = merged.sort_values("score", ascending=False)
sorted_df
```

```python
# value_counts and concat
print(df["department"].value_counts())

more_students = pd.DataFrame({
    "name": ["Farah", "Zoya"],
    "age": [24, 26],
    "department": ["AI", "Data"],
    "score": [95, 60],
})

combined = pd.concat([df, more_students], ignore_index=True)
combined
```

## 4. Data Cleaning

Real-world data is rarely as clean as the table above. Here we build a small, deliberately messy
dataset (missing values, duplicates, inconsistent text, and an outlier) and clean it step by step.


```python
messy_df = pd.DataFrame({
    "name": [" Asha", "ravi", "Meera ", "Kiran", "Kiran", "Farah", None],
    "age": [23, 25, np.nan, 28, 28, 150, 24],       # 150 is an unrealistic outlier
    "department": ["Data", "ai", "Data", "AI", "AI", "Data", "Data"],
    "score": [88, 74, 91, np.nan, 67, 95, 60],
})
messy_df
```

```python
# Step 1: inspect missing values
print(messy_df.isna().sum())
```

```python
# Step 2: handle missing values
clean_df = messy_df.copy()

# Fill missing numeric columns with the column median
clean_df["score"] = clean_df["score"].fillna(clean_df["score"].median())
clean_df["age"] = clean_df["age"].fillna(clean_df["age"].median())

# Drop rows where the name itself is missing, since we cannot recover that
clean_df = clean_df.dropna(subset=["name"])

clean_df
```

```python
# Step 3: remove duplicate rows (Kiran appears twice with identical data)
clean_df = clean_df.drop_duplicates()
clean_df
```

```python
# Step 4: standardise inconsistent text (extra spaces, mixed case)
clean_df["name"] = clean_df["name"].str.strip().str.title()
clean_df["department"] = clean_df["department"].str.strip().str.upper()
clean_df
```

```python
# Step 5: detect and handle outliers using the IQR method
q1 = clean_df["age"].quantile(0.25)
q3 = clean_df["age"].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print("valid age range:", lower_bound, "to", upper_bound)
outliers = clean_df[(clean_df["age"] < lower_bound) | (clean_df["age"] > upper_bound)]
print("outlier rows:\n", outliers)

# Replace the outlier age with the median age instead of dropping the whole row
median_age = clean_df.loc[
    (clean_df["age"] >= lower_bound) & (clean_df["age"] <= upper_bound), "age"
].median()
clean_df.loc[(clean_df["age"] < lower_bound) | (clean_df["age"] > upper_bound), "age"] = median_age
clean_df
```

```python
# Step 6: fix data types
clean_df["age"] = clean_df["age"].astype(int)
clean_df.dtypes
```

## 5. File Handling

Basic reading and writing of text, CSV and JSON files using Python's built-in tools. (Error handling
with `try`/`except` is intentionally left out of this notebook and covered separately.)


```python
# Plain text files
with open("notes.txt", "w") as f:
    f.write("Day 2 covers AI, ML, DL and GenAI foundations.\n")
    f.write("We also build a linear regression and a small neural network.\n")

with open("notes.txt", "r") as f:
    contents = f.read()

print(contents)
```

```python
# CSV files with the csv module
import csv

rows = [
    {"name": "Asha", "score": 88},
    {"name": "Ravi", "score": 74},
    {"name": "Meera", "score": 91},
]

with open("scores.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "score"])
    writer.writeheader()
    writer.writerows(rows)

with open("scores.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row)
```

```python
# JSON files
import json

course_info = {
    "day": 2,
    "topics": ["AI vs ML vs DL vs GenAI", "Linear Regression", "ANN Intuition", "Clustering", "KNN"],
    "duration_hours": 4,
}

with open("course_info.json", "w") as f:
    json.dump(course_info, f, indent=2)

with open("course_info.json", "r") as f:
    loaded_info = json.load(f)

print(loaded_info)
```
