## Core LLM and embedding utilities

These two wrapper functions are the only interface to the OpenAI API used throughout the
notebook. Every technique below is composed from `embed_texts` and `chat`.


```python
def embed_texts(texts: List[str]) -> np.ndarray:
    '''Embed a list of texts with text-embedding-3-small. Returns an (N, D) float32 array,
    L2-normalized so that dot product equals cosine similarity.'''
    if len(texts) == 0:
        return np.zeros((0, 1536), dtype=np.float32)
    resp = client.embeddings.create(model=EMBED_MODEL, input=texts)
    vecs = np.array([d.embedding for d in resp.data], dtype=np.float32)
    norms = np.linalg.norm(vecs, axis=1, keepdims=True)
    norms[norms == 0] = 1e-8
    return vecs / norms


def chat(prompt: str, system: Optional[str] = None, temperature: float = 0.0,
         model: str = CHAT_MODEL, max_tokens: int = 600) -> str:
    '''Single-turn chat completion wrapper used for every LLM-driven step
    (rewriting, expansion, reranking, grading, compression, generation, etc.).'''
    messages = []
    if system:
        messages.append({"role": "system", "content": system})
    messages.append({"role": "user", "content": prompt})
    resp = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens,
    )
    return resp.choices[0].message.content.strip()
```

```python
def split_into_chunks(text: str, max_sentences: int = 2) -> List[str]:
    '''Simple sentence-based chunker used to create child chunks from each parent document.'''
    sentences = re.split(r'(?<=[.!?])\s+', text.strip())
    chunks = []
    for i in range(0, len(sentences), max_sentences):
        chunk = " ".join(sentences[i:i + max_sentences]).strip()
        if chunk:
            chunks.append(chunk)
    return chunks


CHILD_CHUNKS = []  # each: {chunk_id, parent_id, text, category, product_area, updated, title}
for parent in PARENT_DOCS:
    pieces = split_into_chunks(parent["text"], max_sentences=2)
    for i, piece in enumerate(pieces):
        CHILD_CHUNKS.append({
            "chunk_id": f"{parent['doc_id']}-C{i+1}",
            "parent_id": parent["doc_id"],
            "text": piece,
            "category": parent["category"],
            "product_area": parent["product_area"],
            "updated": parent["updated"],
            "title": parent["title"],
        })

PARENT_BY_ID = {p["doc_id"]: p for p in PARENT_DOCS}
print(f"Generated {len(CHILD_CHUNKS)} child chunks from {len(PARENT_DOCS)} parent documents.")
pd.DataFrame(CHILD_CHUNKS)[["chunk_id", "parent_id", "category", "text"]].head(6)
```

## Build the retrieval indexes

Two indexes are built once and reused throughout the notebook:

1. A **dense index**: embeddings of every child chunk using `text-embedding-3-small`.
2. A **sparse index**: a BM25 index over the same child chunks using `rank_bm25`.

Both operate on identical chunk boundaries so that later comparisons between dense, sparse,
and hybrid retrieval are apples-to-apples.


```python
from rank_bm25 import BM25Okapi

CHUNK_TEXTS = [c["text"] for c in CHILD_CHUNKS]

# ---- Dense index ----
print("Embedding", len(CHUNK_TEXTS), "chunks with", EMBED_MODEL, "...")
DENSE_INDEX = embed_texts(CHUNK_TEXTS)
print("Dense index shape:", DENSE_INDEX.shape)

# ---- Sparse (BM25) index ----
def tokenize(text: str) -> List[str]:
    return re.findall(r"[a-z0-9]+", text.lower())

TOKENIZED_CORPUS = [tokenize(t) for t in CHUNK_TEXTS]
BM25_INDEX = BM25Okapi(TOKENIZED_CORPUS)
print("BM25 index built over", len(TOKENIZED_CORPUS), "chunks.")
```

```python
def dense_search(query: str, k: int = 5, allowed_idx: Optional[List[int]] = None) -> List[Tuple[int, float]]:
    '''Dense (embedding) retrieval. Returns list of (chunk_index, score) sorted descending.'''
    q_vec = embed_texts([query])[0]
    if allowed_idx is None:
        scores = DENSE_INDEX @ q_vec
        order = np.argsort(-scores)[:k]
        return [(int(i), float(scores[i])) for i in order]
    else:
        sub_scores = DENSE_INDEX[allowed_idx] @ q_vec
        order = np.argsort(-sub_scores)[:k]
        return [(allowed_idx[i], float(sub_scores[i])) for i in order]


def sparse_search(query: str, k: int = 5, allowed_idx: Optional[List[int]] = None) -> List[Tuple[int, float]]:
    '''Sparse (BM25) retrieval. Returns list of (chunk_index, score) sorted descending.'''
    scores = BM25_INDEX.get_scores(tokenize(query))
    if allowed_idx is not None:
        mask = np.full(len(scores), -np.inf)
        mask[allowed_idx] = scores[allowed_idx]
        scores = mask
    order = np.argsort(-scores)[:k]
    return [(int(i), float(scores[i])) for i in order]


def show_results(results: List[Tuple[int, float]], title: str = ""):
    rows = []
    for idx, score in results:
        c = CHILD_CHUNKS[idx]
        rows.append({"score": round(score, 4), "chunk_id": c["chunk_id"],
                      "category": c["category"], "text": c["text"][:100] + "..."})
    df = pd.DataFrame(rows)
    if title:
        print(title)
    return df
```

# Section 2 — Retrieval Improvements

## 2.1 Dense vs Sparse Retrieval

Dense retrieval (embeddings) captures semantic similarity and handles paraphrase well but can
miss exact keyword / identifier matches (error codes, header names, product-specific terms).
Sparse retrieval (BM25) is the reverse: it excels at exact lexical overlap but misses
paraphrases. The cell below runs the same query through both and highlights where they
disagree.


```python
compare_query = "What HTTP header does NimbusCloud use to verify a webhook payload?"

dense_hits = dense_search(compare_query, k=5)
sparse_hits = sparse_search(compare_query, k=5)

print("DENSE (embedding) results:")
display(show_results(dense_hits))

print("\nSPARSE (BM25) results:")
display(show_results(sparse_hits))

dense_ids = set(i for i, _ in dense_hits)
sparse_ids = set(i for i, _ in sparse_hits)
print("\nOverlap between dense and sparse top-5:", len(dense_ids & sparse_ids), "of 5")
print("This query contains a specific technical term ('X-Nimbus-Signature' header), which is "
      "exactly the case where BM25's exact-token matching can outperform pure embedding "
      "similarity, motivating hybrid search below.")
```

## 2.2 Hybrid Search

Hybrid search combines dense and sparse scores into a single ranking instead of choosing one.
Because dense (cosine, roughly -1..1) and BM25 (unbounded, corpus-dependent) scores live on
different scales, each score list is min-max normalized to [0, 1] before being combined with
a weight `alpha`.


```python
def min_max_norm(scores: np.ndarray) -> np.ndarray:
    finite = scores[np.isfinite(scores)]
    if finite.size == 0 or finite.max() == finite.min():
        return np.zeros_like(scores)
    lo, hi = finite.min(), finite.max()
    out = (scores - lo) / (hi - lo)
    out[~np.isfinite(scores)] = 0.0
    return out


def hybrid_search(query: str, k: int = 5, alpha: float = 0.5) -> List[Tuple[int, float]]:
    '''alpha=1.0 -> pure dense, alpha=0.0 -> pure sparse.'''
    q_vec = embed_texts([query])[0]
    dense_scores = DENSE_INDEX @ q_vec
    sparse_scores = np.array(BM25_INDEX.get_scores(tokenize(query)))

    dense_norm = min_max_norm(dense_scores)
    sparse_norm = min_max_norm(sparse_scores)

    combined = alpha * dense_norm + (1 - alpha) * sparse_norm
    order = np.argsort(-combined)[:k]
    return [(int(i), float(combined[i])) for i in order]


hybrid_hits = hybrid_search(compare_query, k=5, alpha=0.5)
print("HYBRID (alpha=0.5) results:")
display(show_results(hybrid_hits))
```

## 2.3 Reciprocal Rank Fusion (RRF)

RRF is a rank-based (not score-based) fusion method, which avoids the score-scale problem
entirely: each retriever contributes `1 / (k_rrf + rank)` for every document it ranks, and the
contributions are summed across retrievers. This makes RRF robust when combining retrievers
whose raw scores are not directly comparable, including more than two retrievers at once.


```python
def reciprocal_rank_fusion(rank_lists: List[List[int]], k_rrf: int = 60, top_k: int = 5) -> List[Tuple[int, float]]:
    '''rank_lists: list of ranked chunk-index lists, one per retriever (best first).'''
    fused_scores: Dict[int, float] = {}
    for rank_list in rank_lists:
        for rank, idx in enumerate(rank_list):
            fused_scores[idx] = fused_scores.get(idx, 0.0) + 1.0 / (k_rrf + rank + 1)
    ranked = sorted(fused_scores.items(), key=lambda x: -x[1])[:top_k]
    return [(idx, score) for idx, score in ranked]


dense_rank_list = [i for i, _ in dense_search(compare_query, k=15)]
sparse_rank_list = [i for i, _ in sparse_search(compare_query, k=15)]

rrf_hits = reciprocal_rank_fusion([dense_rank_list, sparse_rank_list], k_rrf=60, top_k=5)
print("RRF-fused results (dense + sparse):")
display(show_results(rrf_hits))
```

## 2.4 Re-ranking (LLM-based relevance scoring)

Retrieval with embeddings or BM25 is optimized for recall over a large corpus, not precision
on the final short list. A common industry pattern is to over-retrieve (e.g. top 15) with a
cheap retriever, then re-rank that candidate set with a more expensive but more accurate
scorer before it reaches the generation step. Here `gpt-4o-mini` scores each candidate for
relevance to the query on a 0-10 scale.


```python
def llm_rerank(query: str, candidate_idx: List[int], top_k: int = 5) -> List[Tuple[int, float]]:
    candidates_text = "\n".join(
        f"{i}. {CHILD_CHUNKS[idx]['text']}" for i, idx in enumerate(candidate_idx)
    )
    system = ("You are a precise relevance grader for a retrieval system. Score how relevant "
              "each numbered passage is to the query, from 0 (irrelevant) to 10 (directly "
              "answers the query). Respond ONLY with a JSON array of integers, one score per "
              "passage, in the same order, no other text.")
    prompt = f"Query: {query}\n\nPassages:\n{candidates_text}\n\nJSON array of scores:"
    raw = chat(prompt, system=system, temperature=0.0)
    try:
        scores = json.loads(re.search(r"\[.*\]", raw, re.S).group())
    except Exception:
        scores = [0] * len(candidate_idx)
    scored = list(zip(candidate_idx, scores))
    scored.sort(key=lambda x: -x[1])
    return [(idx, float(s)) for idx, s in scored[:top_k]]


over_retrieved = [i for i, _ in dense_search(compare_query, k=15)]
llm_reranked = llm_rerank(compare_query, over_retrieved, top_k=5)
print("LLM re-ranked (top 15 dense candidates re-scored by gpt-4o-mini):")
display(show_results(llm_reranked))
```

## 2.5 Cross-Encoder Reranker

LLM-based re-ranking is accurate but relatively slow and costly per query. A cross-encoder is
a small transformer trained specifically to score a (query, passage) pair jointly — unlike
the bi-encoder used for dense retrieval, which embeds query and passage independently. Cross-
encoders are the standard industry choice for a fast, self-hosted re-ranking stage.


```python
from sentence_transformers import CrossEncoder

cross_encoder = CrossEncoder("cross-encoder/ms-marco-MiniLM-L-6-v2")

def cross_encoder_rerank(query: str, candidate_idx: List[int], top_k: int = 5) -> List[Tuple[int, float]]:
    pairs = [(query, CHILD_CHUNKS[idx]["text"]) for idx in candidate_idx]
    scores = cross_encoder.predict(pairs)
    scored = list(zip(candidate_idx, scores))
    scored.sort(key=lambda x: -x[1])
    return [(idx, float(s)) for idx, s in scored[:top_k]]


ce_reranked = cross_encoder_rerank(compare_query, over_retrieved, top_k=5)
print("Cross-encoder re-ranked (top 15 dense candidates re-scored by MiniLM cross-encoder):")
display(show_results(ce_reranked))

print("\nComparison of final top-3 chunk_ids across strategies:")
print("Dense only     :", [CHILD_CHUNKS[i]['chunk_id'] for i, _ in dense_hits[:3]])
print("Hybrid         :", [CHILD_CHUNKS[i]['chunk_id'] for i, _ in hybrid_hits[:3]])
print("RRF            :", [CHILD_CHUNKS[i]['chunk_id'] for i, _ in rrf_hits[:3]])
print("LLM rerank     :", [CHILD_CHUNKS[i]['chunk_id'] for i, _ in llm_reranked[:3]])
print("Cross-encoder  :", [CHILD_CHUNKS[i]['chunk_id'] for i, _ in ce_reranked[:3]])
```

# Section 3 — Query Improvements

Retrieval quality is bounded by query quality. This section addresses queries that are
ambiguous, colloquial, compound, or phrased at the wrong level of abstraction for the
knowledge base.


```python
messy_query = "annual plan switch money back thing, can i still get some cash if i cancel early"
```

## 3.1 Query Rewriting

An LLM rewrites an informally phrased or ambiguous user query into a clear, retrieval-friendly
query, correcting grammar and making implicit intent explicit, without changing its meaning.


```python
def rewrite_query(raw_query: str) -> str:
    system = ("Rewrite the user's question into a single clear, well-formed search query for a "
              "SaaS support knowledge base. Preserve the original intent exactly. Do not answer "
              "the question. Return only the rewritten query, no quotes, no preamble.")
    return chat(raw_query, system=system)


rewritten = rewrite_query(messy_query)
print("RAW QUERY     :", messy_query)
print("REWRITTEN     :", rewritten)

print("\nRetrieval with raw query:")
display(show_results(dense_search(messy_query, k=3)))
print("\nRetrieval with rewritten query:")
display(show_results(dense_search(rewritten, k=3)))
```

## 3.2 Query Expansion

Query expansion appends related terms, synonyms, or likely-relevant vocabulary to the
original query before retrieval, increasing the chance of lexical/semantic overlap with the
relevant chunk without changing the query's core meaning. This is especially useful for
sparse and hybrid retrieval.


```python
def expand_query(query: str) -> str:
    system = ("Given a search query for a SaaS support knowledge base, list 5-8 closely related "
              "keywords or phrases (synonyms, related product terms) that could appear in a "
              "relevant document. Return only a comma-separated list, no explanation.")
    expansion = chat(query, system=system)
    return f"{query} {expansion.replace(',', ' ')}"


expanded = expand_query("can I get money back if I cancel my subscription early")
print("EXPANDED QUERY:", expanded)
display(show_results(dense_search(expanded, k=5)))
```

## 3.3 Multi-Query Retrieval

Instead of trusting a single phrasing of the query, an LLM generates several diverse
reformulations, each is retrieved independently, and the unique results are merged. This
increases recall for queries where the "right" phrasing to match the corpus is not obvious in
advance.


```python
def generate_query_variants(query: str, n: int = 4) -> List[str]:
    system = (f"Generate {n} diverse reformulations of the given question for a knowledge-base "
              "search engine. Vary vocabulary and phrasing while preserving intent. Return each "
              "reformulation on its own line, no numbering, no extra text.")
    raw = chat(query, system=system)
    variants = [v.strip() for v in raw.split("\n") if v.strip()]
    return [query] + variants[:n]


def multi_query_retrieve(query: str, k_per_query: int = 4, top_k: int = 5) -> List[Tuple[int, float]]:
    variants = generate_query_variants(query, n=4)
    seen: Dict[int, float] = {}
    for v in variants:
        for idx, score in dense_search(v, k=k_per_query):
            seen[idx] = max(seen.get(idx, -1e9), score)
    ranked = sorted(seen.items(), key=lambda x: -x[1])[:top_k]
    return ranked, variants


mq_query = "will my files still sync if I'm on my phone and my wifi drops"
mq_hits, mq_variants = multi_query_retrieve(mq_query)
print("Original query:", mq_query)
print("Generated variants:")
for v in mq_variants:
    print(" -", v)
print("\nMerged multi-query retrieval results:")
display(show_results(mq_hits))
```

## 3.4 HyDE — Hypothetical Document Embeddings

HyDE flips the retrieval direction: instead of embedding the (often terse) question, the LLM
first writes a hypothetical passage that would answer the question, and that hypothetical
answer is embedded and used for similarity search. Because it is closer in style and content
density to the actual knowledge-base passages, it frequently retrieves better than the raw
question, particularly for short or vague questions.


```python
def hyde_retrieve(query: str, k: int = 5) -> Tuple[List[Tuple[int, float]], str]:
    system = ("Write a short, confident, factual-sounding passage (3-4 sentences) that would "
              "plausibly answer the following support question, in the style of a SaaS "
              "knowledge-base article. It does not need to be accurate for a real product; it "
              "only needs to be representative of what the answer would look like.")
    hypothetical_doc = chat(query, system=system)
    q_vec = embed_texts([hypothetical_doc])[0]
    scores = DENSE_INDEX @ q_vec
    order = np.argsort(-scores)[:k]
    return [(int(i), float(scores[i])) for i in order], hypothetical_doc


hyde_query = "is my data safe if there's a breach"
hyde_hits, hyde_doc = hyde_retrieve(hyde_query)
print("Query:", hyde_query)
print("\nHypothetical document generated by gpt-4o-mini:\n", hyde_doc)
print("\nRetrieval using the hypothetical document's embedding:")
display(show_results(hyde_hits))
```

## 3.5 Query Decomposition

Compound questions that really contain multiple sub-questions retrieve poorly as a single
query, because the single embedding is a blurred average of several distinct information
needs. Decomposition splits the question into independent sub-questions, retrieves for each
separately, and the union of results is passed to generation.


```python
def decompose_query(query: str) -> List[str]:
    system = ("If the following question contains multiple distinct sub-questions, split it "
              "into separate, self-contained sub-questions, one per line, no numbering. If it is "
              "already a single question, return it unchanged on one line.")
    raw = chat(query, system=system)
    return [q.strip() for q in raw.split("\n") if q.strip()]


def decomposed_retrieve(query: str, k_per_sub: int = 3) -> Tuple[List[Tuple[int, float]], List[str]]:
    sub_questions = decompose_query(query)
    seen: Dict[int, float] = {}
    for sq in sub_questions:
        for idx, score in dense_search(sq, k=k_per_sub):
            seen[idx] = max(seen.get(idx, -1e9), score)
    ranked = sorted(seen.items(), key=lambda x: -x[1])
    return ranked, sub_questions


compound_query = "what encryption do you use for stored files, and separately, how do I set up SSO with Okta?"
decomp_hits, sub_questions = decomposed_retrieve(compound_query)
print("Compound query:", compound_query)
print("\nDecomposed sub-questions:")
for sq in sub_questions:
    print(" -", sq)
print("\nUnion of retrieved chunks across sub-questions:")
display(show_results(decomp_hits))
```

## 3.6 Step-back Prompting

Step-back prompting asks the LLM to first generate a more general, higher-level question that
the specific question is an instance of, retrieves using that broader question to surface
foundational/background context, and then answers the original specific question using both
the broad context and (optionally) chunks retrieved for the specific question directly. This
helps when the specific question needs background context that would not itself contain the
specific keywords.


```python
def step_back_retrieve(query: str, k: int = 4) -> Tuple[List[Tuple[int, float]], str]:
    system = ("Given a specific question, write one more general, higher-level 'step-back' "
              "question whose answer would provide useful background context for answering the "
              "specific question. Return only the step-back question.")
    step_back_q = chat(query, system=system)
    step_back_hits = dense_search(step_back_q, k=k)
    direct_hits = dense_search(query, k=k)
    combined = {idx: score for idx, score in step_back_hits}
    for idx, score in direct_hits:
        combined[idx] = max(combined.get(idx, -1e9), score)
    ranked = sorted(combined.items(), key=lambda x: -x[1])
    return ranked, step_back_q


specific_query = "can a compliance admin override the 30-day trash auto-delete for one specific user's workspace?"
sb_hits, step_back_question = step_back_retrieve(specific_query)
print("Specific query:", specific_query)
print("Step-back (general) question:", step_back_question)
print("\nCombined retrieval (step-back context + direct):")
display(show_results(sb_hits))
```

# Section 4 — Context Improvements

Once the right chunks are retrieved, how they are packaged for the generator still matters:
irrelevant sentences waste context budget, position in the prompt affects how well the LLM
uses each chunk, and a chunk's small size for precise retrieval may be too small for
sufficient answer context. This section addresses all three.


## 4.1 Contextual Compression

Retrieved chunks often contain sentences that are not actually relevant to the query, even if
the chunk as a whole scored well. Contextual compression asks an LLM to extract only the
sentences from each retrieved chunk that are relevant to the specific query, shrinking the
context passed to the final generation step and reducing distraction from irrelevant text.


```python
def compress_chunk(query: str, chunk_text: str) -> str:
    system = ("Extract only the sentence(s) from the passage below that are directly relevant "
              "to answering the query. Return the extracted text verbatim with no commentary. "
              "If nothing is relevant, return an empty string.")
    prompt = f"Query: {query}\n\nPassage: {chunk_text}"
    return chat(prompt, system=system, temperature=0.0, max_tokens=150)


compression_query = "what happens if my credit card payment fails"
raw_hits = dense_search(compression_query, k=4)
print("Before compression:")
for idx, _ in raw_hits:
    print(" -", CHILD_CHUNKS[idx]["text"])

print("\nAfter contextual compression:")
compressed_chunks = []
for idx, _ in raw_hits:
    compressed = compress_chunk(compression_query, CHILD_CHUNKS[idx]["text"])
    compressed_chunks.append(compressed)
    print(" -", compressed if compressed else "[filtered out - not relevant]")
```

## 4.2 Chunk Reordering

LLMs are known to attend less reliably to information placed in the middle of a long context
window than to information at the very start or very end (the "lost in the middle" effect).
Chunk reordering keeps the ranking from retrieval but re-arranges the physical order passed
to the prompt so that the most relevant chunks sit at the beginning and end, with less
relevant chunks pushed to the middle.


```python
def reorder_for_lost_in_middle(ranked_idx: List[int]) -> List[int]:
    '''Given chunk indices already sorted best-first, interleave them so the strongest
    chunks land at the start and end of the sequence rather than all at the front.'''
    result = [None] * len(ranked_idx)
    left, right = 0, len(ranked_idx) - 1
    for i, idx in enumerate(ranked_idx):
        if i % 2 == 0:
            result[left] = idx
            left += 1
        else:
            result[right] = idx
            right -= 1
    return result


ranked_for_reorder = [idx for idx, _ in dense_search("storage overage billing rules", k=6)]
print("Original relevance order (best first): ", [CHILD_CHUNKS[i]['chunk_id'] for i in ranked_for_reorder])
reordered = reorder_for_lost_in_middle(ranked_for_reorder)
print("Reordered for prompt (best at edges):  ", [CHILD_CHUNKS[i]['chunk_id'] for i in reordered])
```

## 4.3 Parent-Child Retrieval

Small chunks retrieve precisely (a 1-2 sentence chunk closely matches a narrow query) but may
lack surrounding context needed for a complete answer. Parent-child retrieval resolves this
by retrieving on the small child chunks, then substituting in each child's full parent
document as the context actually sent to the generator.


```python
def parent_child_retrieve(query: str, k: int = 3) -> List[Dict]:
    child_hits = dense_search(query, k=k)
    seen_parents = set()
    results = []
    for idx, score in child_hits:
        child = CHILD_CHUNKS[idx]
        parent_id = child["parent_id"]
        if parent_id in seen_parents:
            continue
        seen_parents.add(parent_id)
        parent = PARENT_BY_ID[parent_id]
        results.append({
            "matched_child": child["text"],
            "score": score,
            "parent_id": parent_id,
            "parent_title": parent["title"],
            "parent_text": parent["text"],
        })
    return results


pc_query = "what's the retry behavior for failed webhook deliveries"
pc_results = parent_child_retrieve(pc_query, k=3)
for r in pc_results:
    print(f"Matched child chunk (score {r['score']:.3f}): {r['matched_child']}")
    print(f"-> Returning full parent document '{r['parent_title']}' for context:")
    print("   " + r["parent_text"])
    print()
```

## 4.4 Metadata Filtering

Retrieval quality improves substantially when structured metadata (category, product area,
recency) is used to narrow the search space before or after semantic ranking, especially when
the question implies a filter ("recently updated", "billing-related", a named category) that
pure text similarity cannot reliably infer. Here an LLM extracts a structured filter from the
natural-language query, which is then applied against the corpus metadata.


```python
VALID_CATEGORIES = sorted(set(c["category"] for c in CHILD_CHUNKS))

def extract_metadata_filter(query: str) -> Dict:
    system = (f"Extract a metadata filter from the user's query for a support knowledge base. "
              f"Valid categories: {VALID_CATEGORIES}. Return a JSON object with optional keys "
              f"'category' (one of the valid categories or null) and 'min_updated' (a date "
              f"string 'YYYY-MM-DD' if the user implies recency, else null). Return JSON only.")
    raw = chat(query, system=system, temperature=0.0)
    try:
        return json.loads(re.search(r"\{.*\}", raw, re.S).group())
    except Exception:
        return {"category": None, "min_updated": None}


def filtered_retrieve(query: str, k: int = 5) -> Tuple[List[Tuple[int, float]], Dict]:
    filt = extract_metadata_filter(query)
    allowed = list(range(len(CHILD_CHUNKS)))
    if filt.get("category"):
        allowed = [i for i in allowed if CHILD_CHUNKS[i]["category"] == filt["category"]]
    if filt.get("min_updated"):
        allowed = [i for i in allowed if CHILD_CHUNKS[i]["updated"] >= filt["min_updated"]]
    hits = dense_search(query, k=k, allowed_idx=allowed if allowed else None)
    return hits, filt


meta_query = "show me the most recently updated security policy about encryption"
meta_hits, extracted_filter = filtered_retrieve(meta_query)
print("Query:", meta_query)
print("Extracted metadata filter:", extracted_filter)
print("\nFiltered retrieval results:")
display(show_results(meta_hits))
```

# Section 6 — Production Pipeline and Comparative Evaluation

## 6.1 Putting it together: `NimbusRAG`

A single class assembling the techniques into configurable stages, representing what a
production deployment would expose behind an internal API: query understanding, retrieval,
context shaping, and a corrective quality gate, all optional so the pipeline can be tuned per
use case (e.g. a cheaper configuration for low-stakes FAQ traffic, a fuller configuration for
compliance-sensitive questions).


```python
class NimbusRAG:
    def __init__(self, use_query_rewrite=True, use_hyde=False, use_hybrid=True,
                 use_cross_encoder_rerank=True, use_compression=True,
                 use_reordering=True, use_corrective_gate=True,
                 wide_k=15, final_k=5):
        self.use_query_rewrite = use_query_rewrite
        self.use_hyde = use_hyde
        self.use_hybrid = use_hybrid
        self.use_cross_encoder_rerank = use_cross_encoder_rerank
        self.use_compression = use_compression
        self.use_reordering = use_reordering
        self.use_corrective_gate = use_corrective_gate
        self.wide_k = wide_k
        self.final_k = final_k

    def run(self, raw_query: str) -> Dict:
        trace = {"raw_query": raw_query}
        query = raw_query

        if self.use_query_rewrite:
            query = rewrite_query(raw_query)
        trace["query_used"] = query

        if self.use_hyde:
            candidates, hyde_doc = hyde_retrieve(query, k=self.wide_k)
            trace["hyde_doc"] = hyde_doc
            candidate_idx = [idx for idx, _ in candidates]
        elif self.use_hybrid:
            candidate_idx = [idx for idx, _ in hybrid_search(query, k=self.wide_k, alpha=0.5)]
        else:
            candidate_idx = [idx for idx, _ in dense_search(query, k=self.wide_k)]
        trace["stage1_candidates"] = [CHILD_CHUNKS[i]["chunk_id"] for i in candidate_idx]

        if self.use_corrective_gate:
            preview = "\n".join(f"- {CHILD_CHUNKS[i]['text']}" for i in candidate_idx[:self.wide_k])
            grade = chat(
                f"Query: {query}\n\nPassages:\n{preview}",
                system=("Classify retrieval quality as exactly one word: CORRECT, INCORRECT, "
                        "or AMBIGUOUS."), temperature=0.0, max_tokens=10).strip().upper()
            trace["corrective_grade"] = grade
            if "INCORRECT" in grade:
                trace["answer"] = ("This knowledge base does not appear to contain information "
                                    "to answer that question.")
                return trace

        if self.use_cross_encoder_rerank:
            final_idx = [idx for idx, _ in cross_encoder_rerank(query, candidate_idx, top_k=self.final_k)]
        else:
            final_idx = candidate_idx[:self.final_k]
        trace["final_chunk_ids"] = [CHILD_CHUNKS[i]["chunk_id"] for i in final_idx]

        if self.use_compression:
            context = [compress_chunk(query, CHILD_CHUNKS[i]["text"]) or CHILD_CHUNKS[i]["text"]
                       for i in final_idx]
        else:
            context = [CHILD_CHUNKS[i]["text"] for i in final_idx]

        if self.use_reordering:
            order = reorder_for_lost_in_middle(list(range(len(context))))
            context = [context[i] for i in order]

        trace["answer"] = generate_answer(raw_query, context)
        return trace


production_pipeline = NimbusRAG()
prod_result = production_pipeline.run(
    "hey so my card got declined twice, what happens to my account and files while its unpaid?"
)
print("Raw query   :", prod_result["raw_query"])
print("Rewritten to:", prod_result["query_used"])
print("Corrective grade:", prod_result.get("corrective_grade"))
print("Final chunks:", prod_result["final_chunk_ids"])
print("\nAnswer:\n", prod_result["answer"])
```

## 6.2 Comparative Evaluation

A lightweight comparison across the pipelines built in this notebook, run on the same set of
representative support questions, to make the practical effect of each layer visible side by
side. `top_chunk_ids` shows which chunk each strategy would place first, which is the most
direct way to see the strategies disagree or converge.


```python
eval_queries = [
    "what happens to my files if I stop paying",
    "how do I connect okta for login",
    "is there a limit on how many api calls i can make per minute",
    "what's the retention window for stuff i delete by accident",
]

rows = []
for q in eval_queries:
    d = dense_search(q, k=1)[0][0]
    h = hybrid_search(q, k=1, alpha=0.5)[0][0]
    rrf = reciprocal_rank_fusion(
        [[i for i, _ in dense_search(q, k=10)], [i for i, _ in sparse_search(q, k=10)]],
        top_k=1)[0][0]
    ce_candidates = [i for i, _ in hybrid_search(q, k=10, alpha=0.5)]
    ce = cross_encoder_rerank(q, ce_candidates, top_k=1)[0][0]
    rows.append({
        "query": q,
        "dense_top1": CHILD_CHUNKS[d]["chunk_id"],
        "hybrid_top1": CHILD_CHUNKS[h]["chunk_id"],
        "rrf_top1": CHILD_CHUNKS[rrf]["chunk_id"],
        "cross_encoder_top1": CHILD_CHUNKS[ce]["chunk_id"],
    })

pd.DataFrame(rows)
```
