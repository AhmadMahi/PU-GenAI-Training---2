```python
%pip install -q requests

import requests

GATEWAY_URL = "https://keygateway1.arshnivlabs.com"
LEARNER_KEY = "pubatch_trainer"

headers = {
    "Authorization": f"Bearer {LEARNER_KEY}",
    "Content-Type": "application/json",
}

# Health
print(requests.get(f"{GATEWAY_URL}/v1/health", timeout=15).status_code)

# Chat — no model (server uses gpt-4o-mini)
chat = requests.post(
    f"{GATEWAY_URL}/v1/chat/completions",
    headers=headers,
    json={"messages": [{"role": "user", "content": "Say hello in one short sentence."}]},
    timeout=30,
)
print(chat.status_code, chat.json()["choices"][0]["message"]["content"])

# Embeddings — no model (server uses text-embedding-3-small)
emb = requests.post(
    f"{GATEWAY_URL}/v1/embeddings",
    headers=headers,
    json={"input": "gateway test"},
    timeout=30,
)
vector = emb.json()["data"][0]["embedding"]
print("EMBEDDING LENGTH:", len(vector))
print("EMBEDDING SAMPLE:", vector[:5])
```
